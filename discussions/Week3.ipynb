# 1. Review of Comparison Operators
Comparison operators are used to compare 2 values. The result of a comparison is always a Boolean value: **True** or **False**. These operators are commonly used inside selection statements to control program behavior.

| Operator | Meaning                  |
| -------- | ------------------------ |
| `==`     | Equal to                 |
| `!=`     | Not equal to             |
| `<`      | Less than                |
| `<=`     | Less than or equal to    |
| `>`      | Greater than             |
| `>=`     | Greater than or equal to |

Check out the simple example code below.

```python
x = 10
y = 5

print(x == y)   # False
print(x != y)   # True
print(x > y)    # True
print(x <= y)   # False
```

Additionally, in Python, comparisons can be chained; this can help with readability. Check out the chain comparisons example below.

```python
z = 20

print(18 <= z < 65)  # True

# equivalent option below
print(z >= 18 and z < 65)
```

## Truth Tables

When combining conditions, we use logical operators. Here is how they decide if the final result is `True` or `False`.

### 1. The `and` Operator
**Rule:** Both sides must be **True** for the result to be True. If even one side is False, the whole thing fails.

| Condition A | Condition B | Result (`A and B`) |
| :---: | :---: | :---: |
| **True** | **True** | ✅ **True** |
| True | False | ❌ False |
| False | True | ❌ False |
| False | False | ❌ False |

---

### 2. The `or` Operator
**Rule:** Only **one** side needs to be True for the result to be True. It only fails if *everything* is False.

| Condition A | Condition B | Result (`A or B`) |
| :---: | :---: | :---: |
| **True** | **True** | ✅ **True** |
| **True** | False | ✅ **True** |
| False | **True** | ✅ **True** |
| False | False | ❌ False |

```python
print(True and False)   # False
print(True or False)    # True
print(not False)        # True
print(not True)         # False
```

Check out the simple example code below.

```python
temperature = 75
is_sunny = True

print((temperature > 70) and is_sunny)  # True
print(False or is_sunny)   # True
print(not True)                   # False
```

# 3. Review of Selection Statements
Selection statements allow a program to choose which code to run based on conditions. Python uses if, elif, and else for decision-making.
Check out the simple if statement example below.

```python
temperature = 32

if True:
    print("Water is frozen.")
```

Check out the if-else statement example below.

```python
temperature = 75

if False:
    print("Water is frozen.")
else:
    print("Water is not frozen.")
```

Check out the if-elif-else statement example below. When there are multiple possible conditions, use elif.

```python
temperature = 212

if temperature <= 32:
    print("Water is frozen.")
elif temperature < 212:
    print("Water is liquid.")
else:
    print("Water is boiling.")
```

## The Difference Between `if` and `elif`

When writing code, you will often need to make choices. A common question is: *“Why can't I just use a bunch of `if` statements? Why do I need `elif`?”*

There are two main reasons: **Logic** and **Efficiency**.

### 1. Logic (Mutual Exclusivity)
*   **Multiple `if` statements:** Every single `if` statement is checked, regardless of what happened before it. If you have three `if` statements and all three conditions are true, all three blocks of code will run.
*   **`elif` statements:** These are connected together. Python only checks an `elif` if the previous `if` (or `elif`) was **False**. As soon as one condition is met, Python executes that block and **skips the rest**.

### 2. Efficiency
*   If you use `elif`, the computer stops working as soon as it finds the answer.
*   If you use multiple `if`s, the computer wastes time checking conditions even after it already found the answer.

```python
# THE WRONG WAY: Using multiple independent 'if' statements
# Scenario: Assigning a letter grade based on a score.

score = 95

print(f"Checking grade for score: {score}...")

if score >= 90:
    print("You got an A!")

if score >= 80:
    print("You got a B!")

if score >= 70:
    print("You got a C!")

if score >= 60:
    print("You passed.")

print("--- Done ---")
```

```python
# ✅ THE RIGHT WAY: Using 'if' and 'elif' (Mutually Exclusive)

score = 95

print(f"Checking grade for score: {score}...")

if score >= 90:
    print("You got an A!")
elif score >= 80:
    # This only runs if score is NOT >= 90
    print("You got a B!")
elif score >= 70:
    # This only runs if score is NOT >= 90 AND NOT >= 80
    print("You got a C!")
elif score >= 60:
    print("You passed.")
else:
    print("You failed.")

print("--- Done ---")
```
