## 🟦 Section 1: Expression Evaluation

Arithmetic Operators
- Python follows PEMDAS order of operations.

| Symbol | Operation | Example |
| :--- | :--- | :--- |
| `+` | Addition | `3 + 2 = 5` |
| `-` | Subtraction | `7 - 4 = 3` |
| `*` | Multiplication | `6 * 2 = 12` |
| `/` | **Float** Division | `5 / 2 = 2.5` |
| `//` | **Integer** Division | `7 // 2 = 3` |
| `%` | Modulus (Remainder)| `7 % 4 = 3` |
| `**` | Exponentiation | `5 ** 2 = 25` |


Comparison Operators
| Operator | Meaning                  |
| -------- | ------------------------ |
| `==`     | Equal to                 |
| `!=`     | Not equal to             |
| `<`      | Less than                |
| `<=`     | Less than or equal to    |
| `>`      | Greater than             |
| `>=`     | Greater than or equal to |
- Float division returns a precise result with a potential fractional part, while integer division returns only the whole number portion, discarding any remainder


Boolean Operators
| Operator | Meaning                                  |
| -------- | ---------------------------------------- |
| `not`    | Reverses the Boolean value               |
| `and`    | True if *both* conditions are true       |
| `or`     | True if *at least one* condition is true |

Precendence order:
1) `not`
2) `and`
3) `or`

Expressions in () are always evaluated first.


- Arithmetic operators are evaluated first, then comparison operators, then boolean operators
- `in` checks membership
- String comparisons (using `==`) are case-sensitive

`````python
'''
print(15/3)
print(15//3)
'''
print(10/3)
print(10//3)

print(15%2)

print( True and True)
print(15 > 5 and 10 < 5)
print(15 > 5 or 10 < 5)

word = 'hi'
print(word == 'HI')
print(word.upper() == 'HI')

print(False or not True and True)
`````

## 🟦 Section 4: For Loops

For loops are used when:
- You know how many times to loop
- You are iterating over a sequence, such as a string or list

Structure:
```python
for i in range(n):
```
```python
for item in list:
```


`````python
# How many times will this print statement be executed?
for i in range(10): # 10 total times
  print(i)
`````

`````python
values = [3, 4, 6]
result = 0

for v in values:
    result = result + v # 5, then 9, then 15

print(result)
`````

## 🟦 Section 6: Nested Loops

A loop inside another loop.

Common uses:
- Grids
- Tables
- Patterns
- performing multiple replicates of the same action, such as repeats of an experiment

`````python
# GO OVER RANDOM FIRST, THEN DO THIS EXAMPLE


# We have 15 students, if they each flip a coin 5 times, how many heads will we get?
import random

random.seed(42)

students = 15
majority_heads = 0

for i in range(students):
    heads = 0
    for j in range(5):
        flip = random.randint(0, 1)  # 0 = tails, 1 = heads
        if flip == 1:
            heads += 1 #heads = heads + 1
    if heads >= 3:
        majority_heads += 1

print(f"Of {students} experiments, {majority_heads} had more heads than tails.")
`````
