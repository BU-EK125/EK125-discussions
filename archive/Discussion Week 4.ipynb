## Part 1: Importing Modules

### What is a Module?
So far, you’ve been using Python’s built-in functions like `print()`, `input()`, `len()`, and `type()`. These are always available. But Python has many more functions organized into modules (also called libraries) that you can access when you need them.

Think of modules as specialized toolboxes:
*   **math** - mathematical functions like square root, trigonometry, constants like π
*   **random** - generating random numbers and making random choices
*   **time** - working with dates, times, and adding delays
*   **os** & **sys** - interacting with your operating system and files

### How to Import: Basic Syntax
To use functions from a module, you must first import it.

**Method 1: Import Specific Functions (Recommended for Beginners)**
`from module_name import function_name`
This imports just the function(s) you need, and you can use them directly.

````python
# Method 1: Import Specific Functions
from random import randint

# Now you can use randint directly
dice_roll = randint(1, 6)
print("You rolled:", dice_roll)

# To import multiple functions from the same module:
from random import randint, random, choice

# Now you can use all three functions
num1 = randint(1, 10)      # Random integer from 1 to 10
num2 = random()            # Random float from 0.0 to 1.0
fruit = choice(['apple', 'banana', 'orange'])  # Random choice from list

print(f"Num1: {num1}, Num2: {num2:.2f}, Fruit: {fruit}")
````

### Method 2: Import the Entire Module
`import module_name`
This imports the whole module, and you use functions by prefixing them with the module name. This is useful to avoid naming conflicts.

### Method 3: Import with an Alias (Nickname)
`import module_name as nickname`
This is common for modules with long names.

**⚠️ Important: Avoid `from module import *`**
This imports everything. It creates a messy namespace and makes debugging harder.

````python
# Method 2: Import Entire Module
import random

# Use the module name before each function
dice_roll = random.randint(1, 6)
probability = random.random()
print("Rolled:", dice_roll)
print("Probability:", probability)

# Method 3: Import with Alias
import random as rnd

dice_roll = rnd.randint(1, 6)
print("Rolled with alias:", dice_roll)
````

## The random Module
Here are the key functions:

*   **`randint(a, b)`**: Generates a random integer from a to b (**inclusive on both ends**).
*   **`random()`**: Generates a random floating-point number from 0.0 (inclusive) to 1.0 (exclusive).

## The math Module
For more advanced math operations like square roots and Pi.

````python
from random import randint, random
from math import sqrt, pi

# --- Random Examples ---
# Roll a six-sided die
roll = randint(1, 6)
print("Rolled:", roll)

# Random decimal between 0 and 100
percentage = random() * 100
print(f"Percentage: {percentage:.2f}%")

# --- Math Examples ---
# Calculate area of a circle
radius = 5
area = pi * radius**2
print(f"Area of circle: {area:.2f}")

# Calculate square root
number = 16
root = sqrt(number)
print(f"Square root of {number} is {root}")
````

 ### Error Checking
Error-checking generally involves continuing to prompt the user until a valid value is entered.

**Note:** When running the code below, you will need to type into the input box that appears.

````python
# Error Checking Example
# Try entering a negative number first to see the loop work!

number = input('Enter a positive number: ')
number = float(number)

while (number <= 0):
    number = input('Seriously! Enter a positive number: ')
    number = float(number)

print('Thanks for entering', number)
````

### Counting in a while loop
It is often useful to count how many times the action ended up being executed. We can use a counter variable initialized before the loop.

````python
counter = 0
number = input('Enter a positive number: ')
number = float(number)
counter = counter + 1

while (number <= 0):
    number = input('Seriously! Enter a positive number: ')
    number = float(number)
    counter = counter + 1

print('Thanks for entering', number)
print('It took you', counter, 'tries.')
````

# Nested Loops

## Nested loops
A nested loop is one loop inside of another.
*   **Outer Loop**: The first loop.
*   **Inner Loop**: The loop inside the action of the outer loop.
The inner loop completes all its iterations for *every single iteration* of the outer loop.

````python
# Nested loop iterating over words and then characters
wordlist = ['hello', "hi", 'ciao']
for myword in wordlist:
    for c in myword:
        print(c, end = ' ')
    print() # Newline after each word
print("That's it!")
'''
print("-" * 20)

# Nested loop with range (Creating a rectangle of stars)
for num in range(3):
    print(f'{num}:', end=' ')
    for n in range(5):
        print('*', end='')
    print()
'''
````

### Varying the Inner Loop
We can make the inner loop depend on the outer loop variable to create shapes (like triangles).

We can also nest `while` loops inside `for` loops (e.g., asking for input 3 distinct times, with error checking for each).

````python
print("Triangle Pattern:")
for num in range(3):
    print(f'{num}:', end=' ')
    for n in range(num+1):
        print('*', end='')
    print()

print("-" * 20)
````

````python
# Nested For/While Loop (Input validation repeated 3 times)
for i in range(3):
    number = input(f'Round {i+1}: Enter a positive number: ')
    number = float(number)
    while (number <= 0):
        print('Please follow directions!')
        number = input('Enter a positive number: ')
        number = float(number)
    print('Thanks for entering', number)
    print()
````

## Using if Inside a Loop
In practice, we often want the action to change depending on the current value of the loop variable.

## Randomness and Reproducibility
In engineering, results must be repeatable.
*   **Pseudo-random numbers**: Generated by a mathematical formula.
*   **Seed**: A starting point for the formula. If you set the seed, you get the same random numbers every time.

````python
# Using IF inside a LOOP (Even/Odd check)
print("Even/Odd Check:")
for n in range(5):
    if n % 2 == 0:
        print(n, "is even")
    else:
        print(n, "is odd")

    if n < 2:
      break

x = 0
while x < 3:
  if x % 2 == 0:
    print(x, "is even")
  else:
    print(x, "is odd")
  x = x + 1
print("-" * 20)

# Randomness and Reproducibility
import random

random.seed(42)  # Set seed

print("Run 1 (Seed 42):")
print(random.randint(1, 100))
print(random.randint(1, 100))

# Resetting seed to show reproducibility
random.seed(42)
print("Run 2 (Seed 42 again):")
print(random.randint(1, 100)) # Will match the first number above
print(random.randint(1, 100)) # Will match the second number above
````
